```python
# [

#   [0, 1, 2, 3, 4],

#   [9, 8, 7, 6, 5],

#   [10, 11, 12, 13, 14]

#   [19, 18, 17, 16, 15]

# ]

row, col = 4, 5
result = [[0] * col for i in range(row)]

start = 8
for i in range(row):
    for j in range(col):
        if i % 2 == 0:
            result[i][j] = col * i + j + start
        else:
            result[i][j] = (col * i + (col - 1)) - j + start
print(result)
```
